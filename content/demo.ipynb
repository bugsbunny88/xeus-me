```python
import importlib
import importlib.metadata
import warnings

# Suppress warnings
warnings.filterwarnings("ignore", category=SyntaxWarning)

# Collect all installed distributions and their versions into a list
dists = []
for dist in importlib.metadata.distributions():
    try:
        name = dist.metadata["Name"]
        version = dist.version
        dists.append((name, version))
    except KeyError:
        # Handle distributions without 'Name' or 'Version' metadata
        pass

# Filter the list to include only packages with import statements available
importable_dists = []
for name, version in dists:
    try:
        importlib.import_module(name)
        importable_dists.append((name, version))
    except (
        ImportError,
        AttributeError,
        TypeError,
        ModuleNotFoundError,
        FileNotFoundError,
        NotImplementedError,
        Exception,
    ):
        pass

# Sort the list alphabetically by package name
sorted_importable_dists = sorted(importable_dists, key=lambda x: x[0].lower())

# Print the sorted list
for name, version in sorted_importable_dists:
    print(f"{name} {version}")
```

```python
# If-elif-else statement
blood_oxygen_saturation = 92.3
altitude = 0

if blood_oxygen_saturation > 96 and altitude == 0:
    print("Healthy individual at sea level")
elif blood_oxygen_saturation > 92 and altitude >= 1000:
    print("Healthy value above sea level")
elif blood_oxygen_saturation == 0:
    print("Monitor failure")
else:
    print("Not good")

# For loop
for base in 'ACTGATGCYGGCA':
    print(base)

# While loop
i = 0
while i < 5:
    print(i)
    i += 1
```

## python-dateutil

### 1. Parsing Dates from Strings

**python-dateutil (2.8.2)**: Provides powerful extensions to the standard `datetime` module.
One of the most common uses of python-dateutil is parsing dates from strings, especially when the format is not fixed or known ahead of time.

```python
from dateutil import parser

date_str1 = "2024-04-01"
date_str2 = "April 1st, 2024 10:00 AM"

date1 = parser.parse(date_str1)
date2 = parser.parse(date_str2)

print(date1)  # 2024-04-01 00:00:00
print(date2)  # 2024-04-01 10:00:00
```

### 2. Relative Deltas

Computing differences between two dates and even adding or subtracting specific time periods from a given date is straightforward with python-dateutil.

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

now = datetime.now()
two_months_later = now + relativedelta(months=+2)
three_weeks_ago = now - relativedelta(weeks=3)

print(f"Current time: {now}")
print(f"Two months from now: {two_months_later}")
print(f"Three weeks ago: {three_weeks_ago}")
```

### 3. Handling Time Zones

python-dateutil provides robust support for time zones, including daylight saving time transitions.

```python
from datetime import datetime
from dateutil import tz

# Define time zones
eastern = tz.gettz('America/New_York')
london = tz.gettz('Europe/London')

# Current time in UTC
utc_now = datetime.utcnow()

# Convert UTC to different time zones
utc_now = utc_now.replace(tzinfo=tz.tzutc())
eastern_now = utc_now.astimezone(eastern)
london_now = utc_now.astimezone(london)

print(f"UTC Now: {utc_now}")
print(f"Eastern Time: {eastern_now}")
print(f"London Time: {london_now}")
```

### 4. Recurring Events with rrule

python-dateutil also supports defining recurring events using rrule.

```python
from dateutil.rrule import rrule, MONTHLY
from dateutil.parser import parse

start_date = parse("2024-01-01")
end_date = parse("2024-12-31")

# Define a monthly recurring event starting from start_date
for dt in rrule(freq=MONTHLY, dtstart=start_date, until=end_date):
    print(dt)
```

## Traitlets Example

Here's a simple example where traitlets is used to create a configurable class. This can be particularly useful in setting up models or analyses with varying parameters.

```python
from traitlets import HasTraits, Int, Float

class Analysis(HasTraits):
    sample_size = Int(100, help="Size of the sample").tag(config=True)
    dropout_rate = Float(0.5, help="Rate of dropout").tag(config=True)

# Example usage
analysis = Analysis()
print(f"Initial sample size: {analysis.sample_size}")
analysis.sample_size = 200
print(f"Updated sample size: {analysis.sample_size}")
```

## Decorator Example

This example demonstrates the use of the decorator package to measure the execution time of a function, which is a common pattern in data science for optimizing code.

```python
from decorator import decorator
import time

@decorator
def timeit(func, *args, **kwargs):
    start = time.time()
    result = func(*args, **kwargs)
    end = time.time()
    print(f"{func.__name__} took {end-start} seconds to run.")
    return result

# Example usage
@timeit
def compute():
    return sum(x*x for x in range(10000))

compute()
```

**decorator (5.1.1)**: Simplifies the usage of decorators for the average programmer and provides a simple way to define signature-preserving decorators.

```python
from decorator import decorator

@decorator
def do_twice(func, *args, **kwargs):
    func(*args, **kwargs)
    return func(*args, **kwargs)

@do_twice
def say_hello(name):
    print(f"Hello {name}")

say_hello("Alice")
```

## pyparsing example

**pyparsing (3.1.1)**: A library for building recursive descent parsers without the need to learn a new syntax. pyparsing allows for creating and executing simple grammars. It's helpful for parsing custom data formats in data science.

```python
from pyparsing import Word, alphas, nums

# Define a simple grammar for a "name age" format
grammar = Word(alphas) + Word(nums)
parsed = grammar.parseString("Alice 42")

name, age = parsed
print(f"Name: {name}, Age: {age}")
```

## cycler example

**cycler (0.12.1)**: A utility for cycling through properties in matplotlib plots. It's useful when you need to create plots with varying properties (like colors, line styles, etc.). cycler is used to create cycles of things like colors or line styles, often for plotting purposes.

```python
from cycler import cycler
import matplotlib.pyplot as plt
import numpy as np

# Create a cycler for colors
color_cycler = cycler('color', ['red', 'green', 'blue'])

plt.rc('axes', prop_cycle=color_cycler)

# Plotting three sine waves with different colors
x = np.linspace(0, 10, 100)
for i in range(3):
    plt.plot(x, np.sin(x + i * np.pi / 3))
plt.show()
```

## attrs example

- **attrs**
    
    - Provides attributes for Python classes, enhancing code readability and maintainability.
    - Import statement: `import attr`
    - **attrs (23.2.0)**: A library that makes defining classes with attributes simpler and more Pythonic.
    - attrs provides a way to define classes without boilerplate. It's useful for creating clean data models.

```python
import attr

@attr.s
class Person:
    name = attr.ib(type=str)
    age = attr.ib(type=int)

# Create a Person instance
person = Person(name="Bob", age=29)
print(f"{person.name} is {person.age} years old.")
```

## Numeric widgets

There are many widgets distributed with ipywidgets that are designed to display numeric values.  Widgets exist for displaying integers and floats, both bounded and unbounded.  The integer widgets share a similar naming scheme to their floating point counterparts.  By replacing `Float` with `Int` in the widget name, you can find the Integer equivalent.

### IntSlider  
- The slider is displayed with a specified, initial `value`. Lower and upper bounds are defined by `min` and `max`, and the value can be incremented according to the `step` parameter.
- The slider's label is defined by `description` parameter 
- The slider's `orientation` is either 'horizontal' (default) or 'vertical'
- `readout`  displays the current value of the slider next to it. The options are **True** (default) or **False** 
    - `readout_format` specifies the format function used to represent slider value. The default is '.2f'
  

```python
widgets.IntSlider(
    value=7,
    min=0,
    max=10,
    step=1,
    description='Test:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True,
    readout_format='d'
)
```
