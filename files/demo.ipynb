```python
import importlib
import importlib.metadata
import warnings

# Suppress warnings
warnings.filterwarnings("ignore", category=SyntaxWarning)

def get_installed_distributions():
    """Fetch all installed distributions with valid names and versions."""
    dists = []
    for dist in importlib.metadata.distributions():
        try:
            name = dist.metadata["Name"]
            version = dist.version
            if version != "0.0.0":  # Skip packages with version '0.0.0'
                dists.append((name, version))
        except KeyError:
            # Handle distributions without 'Name' or 'Version' metadata
            continue
    return dists

def filter_importable_distributions(dists):
    """Filter the list to include only packages that can be imported."""
    importable_dists = []
    for name, version in dists:
        try:
            importlib.import_module(name)
            importable_dists.append((name, version))
        except (
            ImportError,
            AttributeError,
            TypeError,
            ModuleNotFoundError,
            FileNotFoundError,
            NotImplementedError,
            Exception,
        ):
            continue
    return importable_dists

def sort_distributions(distributions):
    """Sort the distributions list alphabetically by package name."""
    return sorted(distributions, key=lambda x: x[0].lower())

def display_distributions(distributions):
    """Display the sorted list of importable distributions."""
    for name, version in distributions:
        print(f"{name} {version}")

# Execution within the notebook
dists = get_installed_distributions()  # Step 1: Get all installed distributions
importable_dists = filter_importable_distributions(dists)  # Step 2: Filter out non-importable distributions
sorted_importable_dists = sort_distributions(importable_dists)  # Step 3: Sort the distributions alphabetically
display_distributions(sorted_importable_dists)  # Step 4: Display the sorted list
```

## 8. Data Analysis with Pandas

**pandas (1.5.3)**: A library providing high-performance, easy-to-use data structures, and data analysis tools. It's commonly used for data manipulation and analysis. Import statement: `import pandas as pd`

```python
import pandas as pd
import numpy as np
from IPython.display import display, Markdown

# ── global display tweaks ──────────────────────────────────────────
pd.set_option("display.max_columns", 15)
pd.set_option("display.max_rows",    10)
pd.set_option("display.precision",    3)

# ── numbered section helper ────────────────────────────────────────
step_counter = 1
def show(text: str):
    """Display a numbered markdown heading."""
    global step_counter
    display(Markdown(f"### {step_counter}. {text}"))
    step_counter += 1

# ==================================================================
# PART 1 ─ demo DataFrame `data`  (A, B, C)
# ==================================================================
np.random.seed(0)
data = pd.DataFrame({
    "A": np.random.randn(100),
    "B": np.random.rand(100),
    "C": np.random.choice(["X", "Y", "Z"], 100),
})
data.loc[[3, 7], "A"] = np.nan               # add a couple of NaNs

show("data – head()")
display(data.head())

show("data – describe()")
display(data.describe())

show("data – info()")
print(data.info(), "\n")

show("data – dropna()")
display(data.dropna())

show("data – fillna(0)")
display(data.fillna(0))

show("data – mean / median / std / max / min")
agg_tbl = data.select_dtypes("number").agg(
    ["mean", "median", "std", "max", "min"]
).T
display(agg_tbl)

show("data – groupby('C').mean()")
display(data.groupby("C").mean(numeric_only=True))

show("data – sort_values(by='B') (first 5)")
display(data.sort_values(by="B").head())

show("data – concat() with new df (first 5)")
df2 = pd.DataFrame(np.random.randn(100, 3), columns=["A", "B", "C"])
display(pd.concat([data, df2], ignore_index=True).head())

show("data – merge() with lookup table")
df3 = pd.DataFrame({"C": ["X", "Y", "Z"], "D": [1, 2, 3]})
display(pd.merge(data, df3, on="C").head())

show("data – pivot_table(mean A by C)")
display(data.pivot_table(values="A", index="C", aggfunc="mean"))

# ==================================================================
# PART 2 ─ large integer DataFrame `random_data_df`
# ==================================================================
np.random.seed(2021)
random_data_df = pd.DataFrame(np.random.randint(1, 25, size=(50, 70)))

show("random_data_df – basic metadata")
print(f"Type  : {type(random_data_df)}")
print(f"Shape : {random_data_df.shape}\n")

show("random_data_df – dtypes")
display(random_data_df.dtypes.to_frame("dtype").T)

show("random_data_df – head()")
display(random_data_df.head())

show("random_data_df – describe()")
display(random_data_df.describe())

first_val  = random_data_df.iat[0, 0]
middle_val = random_data_df.iat[25, 35]
show("random_data_df – specific elements")
print(f"First value  : {first_val}")
print(f"Middle value : {middle_val}\n")

show("random_data_df – slice (first 4 rows × 10 cols)")
display(random_data_df.iloc[:4, :10])

show("random_data_df – overall stats")
overall = pd.Series({
    "Mean": random_data_df.values.mean(),
    "Max" : random_data_df.values.max(),
    "Min" : random_data_df.values.min(),
    "Std" : random_data_df.values.std()
}).to_frame("value").T
display(overall)

row_mean = random_data_df.mean(axis=1).to_frame("row_mean")
col_mean = random_data_df.mean(axis=0).to_frame("col_mean").T
show("random_data_df – row means (first 5)")
display(row_mean.head())

show("random_data_df – column means")
display(col_mean)

# ==================================================================
# PART 3 ─ stacking examples (A, B, C)
# ==================================================================
show("Stacking examples")
A = pd.DataFrame([[1, 2, 3],
                  [4, 5, 6],
                  [7, 8, 9]])

B = pd.concat([A, A], axis=1, keys=["Left", "Right"])
C = pd.concat([A, A], axis=0, ignore_index=True)

display(Markdown("**Horizontal stack (B)**")); display(B)
display(Markdown("**Vertical stack (C)**"));   display(C)
```

## 9. Data Visualization with seaborn

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# demo data
np.random.seed(0)
data = pd.DataFrame({
    "total_bill": np.random.uniform(10, 50, 100),
    "tip":        np.random.uniform(1, 10, 100),
    "sex":        np.random.choice(["Male", "Female"], 100),
    "smoker":     np.random.choice(["Yes", "No"], 100),
    "day":        np.random.choice(["Thur", "Fri", "Sat", "Sun"], 100),
    "time":       np.random.choice(["Lunch", "Dinner"], 100),
})

BASE_PALETTE = ["#c8102e", "#004c97", "#00a44f", "#f2a900"]
sns.set_theme(style="darkgrid")

# pixels to inches at 100 dpi
FIG_W, FIG_H, DPI = 7, 4, 100          # 7×4 in at 100 dpi → 700×400 px

def build_palette(categories):
    return {cat: BASE_PALETTE[i % len(BASE_PALETTE)]
            for i, cat in enumerate(categories)}

# widgets
slider = widgets.IntSlider(
    value=50, min=10, max=50, step=5,
    description="Max total bill:", continuous_update=False
)
hue_dropdown = widgets.Dropdown(
    options=["sex", "smoker", "day", "time"],
    value="sex", description="Hue:"
)
plot_type_radio = widgets.RadioButtons(
    options=["scatterplot", "barplot", "lineplot"],
    value="scatterplot", description="Plot type:"
)

def plot_graph(max_bill, hue, plot_type):
    df = data[data["total_bill"] <= max_bill]
    plt.figure(figsize=(FIG_W, FIG_H), dpi=DPI)     # exact 700 × 400 px

    palette = build_palette(df[hue].unique())

    if plot_type == "scatterplot":
        sns.scatterplot(data=df, x="total_bill", y="tip",
                        hue=hue, palette=palette)

    elif plot_type == "barplot":
        bar_palette = build_palette(df["day"].unique())
        sns.barplot(data=df, x="day", y="total_bill",
                    hue="day", estimator="mean",
                    palette=bar_palette, legend=False)

    elif plot_type == "lineplot":
        sns.lineplot(data=df.sort_values("total_bill"),
                     x="total_bill", y="tip",
                     hue=hue, palette=palette)

    plt.title(f"{plot_type.capitalize()} (total bill ≤ {max_bill})")
    plt.tight_layout()
    plt.show()

def update_plot(change=None):
    clear_output(wait=True)
    display(slider, hue_dropdown, plot_type_radio)
    plot_graph(slider.value, hue_dropdown.value, plot_type_radio.value)

for w in (slider, hue_dropdown, plot_type_radio):
    w.observe(update_plot, names="value")

update_plot()   # initial display
```

## python-dateutil

### 1. Parsing Dates from Strings

**python-dateutil (2.8.2)**: Provides powerful extensions to the standard `datetime` module.
One of the most common uses of python-dateutil is parsing dates from strings, especially when the format is not fixed or known ahead of time.

```python
from dateutil import parser

date_str1 = "2024-04-01"
date_str2 = "April 1st, 2024 10:00 AM"

date1 = parser.parse(date_str1)
date2 = parser.parse(date_str2)

print(date1)  # 2024-04-01 00:00:00
print(date2)  # 2024-04-01 10:00:00
```

### 2. Relative Deltas

Computing differences between two dates and even adding or subtracting specific time periods from a given date is straightforward with python-dateutil.

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

now = datetime.now()
two_months_later = now + relativedelta(months=+2)
three_weeks_ago = now - relativedelta(weeks=3)

print(f"Current time: {now}")
print(f"Two months from now: {two_months_later}")
print(f"Three weeks ago: {three_weeks_ago}")
```

### 3. Handling Time Zones

python-dateutil provides robust support for time zones, including daylight saving time transitions.

```python
from datetime import datetime
from dateutil import tz

# Define time zones
eastern = tz.gettz('America/New_York')
london = tz.gettz('Europe/London')

# Current time in UTC
utc_now = datetime.utcnow()

# Convert UTC to different time zones
utc_now = utc_now.replace(tzinfo=tz.tzutc())
eastern_now = utc_now.astimezone(eastern)
london_now = utc_now.astimezone(london)

print(f"UTC Now: {utc_now}")
print(f"Eastern Time: {eastern_now}")
print(f"London Time: {london_now}")
```

### 4. Recurring Events with rrule

python-dateutil also supports defining recurring events using rrule.

```python
from dateutil.rrule import rrule, MONTHLY
from dateutil.parser import parse

start_date = parse("2024-01-01")
end_date = parse("2024-12-31")

# Define a monthly recurring event starting from start_date
for dt in rrule(freq=MONTHLY, dtstart=start_date, until=end_date):
    print(dt)
```

## Traitlets Example

Here's a simple example where traitlets is used to create a configurable class. This can be particularly useful in setting up models or analyses with varying parameters.

```python
from traitlets import HasTraits, Int, Float

class Analysis(HasTraits):
    sample_size = Int(100, help="Size of the sample").tag(config=True)
    dropout_rate = Float(0.5, help="Rate of dropout").tag(config=True)

# Example usage
analysis = Analysis()
print(f"Initial sample size: {analysis.sample_size}")
analysis.sample_size = 200
print(f"Updated sample size: {analysis.sample_size}")
```

## decorator Example

**decorator (5.1.1)**: Simplifies the usage of decorators for the average programmer and provides a simple way to define signature-preserving decorators. This example demonstrates the use of the decorator package to measure the execution time of a function, which is a common pattern in data science for optimizing code.

```python
from decorator import decorator
import time

@decorator
def timeit(func, *args, **kwargs):
    start = time.time()
    result = func(*args, **kwargs)
    end = time.time()
    print(f"{func.__name__} took {end-start} seconds to run.")
    return result

# Example usage
@timeit
def compute():
    return sum(x*x for x in range(10000))

compute()
```

## requests example

```python
import requests
import pandas as pd
from io import StringIO

# URL of the dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"

# Send a GET request
response = requests.get(url)

# Make sure the request was successful
if response.status_code == 200:
    data = StringIO(response.text)
    tips = pd.read_csv(data)
else:
    print("Failed to retrieve data:", response.status_code)

# Display the first few rows of the DataFrame
print(tips.head())
```
